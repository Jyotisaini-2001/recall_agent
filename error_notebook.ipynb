## 12. Logic Error — Off-by-one in loop (no exception, wrong output)

```python
# Intended: sum numbers 1 through 10
total = 0
for i in range(10):  # Bug: should be range(1, 11)
    total += i

print(f'Sum of 1 to 10 = {total}')  # Prints 45, correct is 55
```

## 13. UnboundLocalError — Referencing local var before assignment

```python
score = 100

def update_score():
    score += 10  # Tries to modify global without declaring it global
    return score

print(update_score())
```
