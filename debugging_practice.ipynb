## Error 1: `ValueError`
**Scenario:** Parsing corrupted CSV data

The sales data contains a non-numeric entry `"N/A"`. `int()` cannot convert it and raises a `ValueError`.

> **Your task:** Fix the loop so it skips (or handles) invalid entries and still prints the correct total.

```python
# BUGGY CODE — run it to see the error
sales_data = ["1200", "950", "N/A", "870"]

total_sales = 0

for value in sales_data:
    total_sales += int(value)   # ValueError on "N/A"

print("Total Sales:", total_sales)
```

```python
# SkillForge fix
sales_data = ["1200", "950", "N/A", "870"]

total_sales = 0

for value in sales_data:
    if value.strip().upper() not in ('N/A', 'NA', '', 'NULL', 'NONE'):
        total_sales += int(value)

print("Total Sales:", total_sales)
```

```python
# SkillForge fix
sales_data = ["1200", "950", "N/A", "870"]

total_sales = 0

for value in sales_data:
    if value.strip().lstrip('-').isdigit():
        total_sales += int(value)

print("Total Sales:", total_sales)
```

```python
# SOLUTION
sales_data = ["1200", "950", "N/A", "870"]

total_sales = 0

for value in sales_data:
    try:
        total_sales += int(value)
    except ValueError:
        print(f"Skipping invalid entry: {value!r}")

print("Total Sales:", total_sales)
```

## Error 4: `ZeroDivisionError`
**Scenario:** Calculating average order value

`orders` is an empty list, so `len(orders)` is `0` — dividing by zero raises `ZeroDivisionError`.

> **Your task:** Guard against an empty list and print a sensible message instead of crashing.

```python
# SkillForge fix
sorted_scores = sorted(
    user_scores,
    key=lambda x: x["score"] if x["score"] is not None else -1
)

print(sorted_scores)
```

```python
# BUGGY CODE — run it to see the error
orders = []

total_amount = sum(order["amount"] for order in orders)

average_order_value = total_amount / len(orders)   # ZeroDivisionError

print("Average Order Value:", average_order_value)
```

```python
# SOLUTION
orders = []

if not orders:
    print("No orders to calculate average.")
else:
    total_amount = sum(order["amount"] for order in orders)
    average_order_value = total_amount / len(orders)
    print("Average Order Value:", average_order_value)
```

## Error 5: `TypeError`
**Scenario:** Sorting mixed API response data

Priya's score is `None`. Python cannot compare `int` and `NoneType` during sort, raising a `TypeError`.

> **Your task:** Fix the sort so `None` scores are handled — treat them as `0` or place them last.

```python
# BUGGY CODE — run it to see the error
user_scores = [
    {"user": "Aman",  "score": 82},
    {"user": "Priya", "score": None},
    {"user": "Ravi",  "score": 91}
]

sorted_scores = sorted(
    user_scores,
    key=lambda x: x["score"]   # TypeError: None vs int
)

print(sorted_scores)
```

```python
# SOLUTION — treat None as -1 so it sorts to the bottom
user_scores = [
    {"user": "Aman",  "score": 82},
    {"user": "Priya", "score": None},
    {"user": "Ravi",  "score": 91}
]

sorted_scores = sorted(
    user_scores,
    key=lambda x: x["score"] if x["score"] is not None else -1
)

for entry in sorted_scores:
    score_display = entry["score"] if entry["score"] is not None else "N/A"
    print(f"{entry['user']}: {score_display}")
```
